# Field Boundary Comparison

Compare **Delineate Anything**, **DelAny-S**, and **FTW** on the 20×20 km AOI subset (Kazakhstan, tile T44TNS).

**Visual QA checklist:**
1. Field corners sharp?
2. Adjacent fields separated?
3. Small fields (0.5–2 ha) preserved?
4. Roads/rivers excluded?
5. Edges align with visible crop/soil lines?

```python
import json
from pathlib import Path
from IPython.display import Image, display

ROOT = Path('..').resolve()
COMPARE = ROOT / 'outputs' / 'compare'

metrics = json.loads((COMPARE / 'metrics.json').read_text())
for m in metrics:
    print(f"{m['model']}: {m['polygon_count']} polygons, "
          f"mean {m.get('mean_area_ha', 0):.1f} ha, "
          f"edge alignment {m.get('edge_alignment', 'n/a')}")
```

```python
display(Image(filename=str(COMPARE / 'side_by_side.png')))
```

```python
for name in ['delineate_anything_overlay.png', 'delineate_anything-s_overlay.png', 'ftw_prue_b5_overlay.png']:
    print(name)
    display(Image(filename=str(COMPARE / name)))
```
